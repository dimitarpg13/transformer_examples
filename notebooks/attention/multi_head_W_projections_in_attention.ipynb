### Multi-Head Projections of the matrices $W_K$, $W_V$, $W_Q$ in the Transformer Attention Block

**Learnable parameters per head** $i \in 1,2,...,h$:

$$W^{(i)}_Q, W^{(i)}_K, W^{(i)}_V \in \mathbb{R}^{d \times d_k}$$


The following self-contained script constructs $h$ independent per-head projection matrices, applies them the naive way, and then stacks them into a fused $(d, d)$ matrix that reproduces the same result via one matmul plus a view.

**single output-mixing matrix**:

$$W_O \in \mathbb{R}^{d \times d}$$

In practice, the $h$ per-head matrices are stacked side-by-side into single $d \times d$ matrices $W_Q, W_K, W_V$ so that one fused matmul produces all heads at once. The "split into heads" then becomes a tensor reshape, not a separate computation.

**per-head projection**:

$$Q^{(i)} = H W^{(i)}_Q,  K^{(i)} = H W^{(i)}_K, V^{(i)} = H W^{(i)}_V, Q^{(i)}, K^{(i)}, V^{(i)} \in \mathbb{R}^{T \times d_k}$$

**row-by-row for token $t$**:

$$q^{(i)}_t = h_t W^{(i)}_Q,  k^{(i)}_t = h_t W^{(i)}_K, v^{(i)}_t = h_t W^{(i)}_V \in \mathbb{R}^{d_k}$$

**cannonical reading of the quantities** $q^{(i)}_t$, $k^{(i)}_t$, $v^{(i)}_t$:

- $q^{(i)}_t$ - **query**: encodes what token $t$ is looking for in the context.

- $k^{(i)}_t$ - **key**: encodes what token $t$ advertises to other positions.

- $v^{(i)}_t$ - **value**: encodes what token t contributes when attended to.
